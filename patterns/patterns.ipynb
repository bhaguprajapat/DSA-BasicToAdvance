HOLLOW RECTANGLE pattern

```python
n= int(input("Enter a number:"))
def Hollow_rectangle(n):
    for i in range(n):
        if i!=0 and i!=n-1:
            print("*"+" " *(n-2)+"*")
        else:
            print('*'*n)
Hollow_rectangle(n)

```

```python
row= int(input("Enter a number:"))
col= int(input("Enter a number:"))
def Hollow_rectangle(row,col):
    for r in range(1,row+1):
        for c in range(1,col+1):
            if r==1 or c==1 or r==row or c==col:
               print("* ",end="")
            else:
               print("  ",end="")
        print("\n")
Hollow_rectangle(row,col)

```

INVERTED & ROTATED HALF-PYRAMID pattern

```python
n= int(input("Enter a number:"))
# n=4
def InnertedRotatedPattern(n):
    for i in range(n):
        print(" "*(n-i)+"*"*(i+1))
InnertedRotatedPattern(n)

```

```python
n=int(input("Enter a number"))
def InvertedHalfPyramid_withNumber(n):
    for i in range(n):
        for j in range(1,n-i+1):
            print(j ,end=" ")
        print("\n")
InvertedHalfPyramid_withNumber(n)
```

```python
n=int(input("Enter a number"))
def floyds_trangle(n):
    counter=1
    for i in range(1,n+1):
        for j in range(1,i):
            print(counter ,end=" ")
            counter+=1
        print("\n")
floyds_trangle(n)
```
