HOLLOW RECTANGLE pattern

```python
n= int(input("Enter a number:"))
def Hollow_rectangle(n):
    for i in range(n):
        if i!=0 and i!=n-1:
            print("*"," " *(n-4),"*")
        else:
            print('*'*n)
Hollow_rectangle(n)

```

INVERTED & ROTATED HALF-PYRAMID pattern

```python
# n= int(input("Enter a number:"))
n=4
def InnertedRotatedPattern(n):
    for i in range(n):
        print(" "*(n-i),"*"*(i+1))
InnertedRotatedPattern(n)

```
