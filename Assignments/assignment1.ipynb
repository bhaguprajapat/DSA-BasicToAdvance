#### Question 1:Given an integer array nums, return true if any value appears at least twice in the array, and return false if every element is distinct.

- Example 1:Input: `nums = [1, 2, 3, 1]` Output:   `true`
- Example 2:Input: `nums = [1, 2, 3, 4]` Output:   `false `
- Example 3:Input:`nums = [1, 1, 1, 3, 3, 4, 3, 2, 4, 2]` Output:   `true`
 Constraints: `•1 <= nums .  lengtth <= 105•-109 <= nums [ i ] <= 109`

# Approach
Brute force approach
first I sort the array
then run the loop
and check all element if any one element match with last element return true otherwise return false

```python
def containsDuplicate(nums):
    nums.sort()
    for i in range(1,len(nums)):
        if(nums[i]==nums[i-1]):
            return True
    return False
nums=[1,2,3,5]
print(containsDuplicate(nums))
```

#### Question 2:
There is an integer array nums sorted in ascending order (with distinct values).Prior to being passed to your function, nums is possibly rotated at an unknown pivot  index `k (1  <=  k  <  nums.length)`  such  that  the  resulting  array  is [`nums[k], nums[k+1],`   ...,   `nums[n-1], nums[0],   nums[1],   ...,   nums[k-1]] (0-indexed).`   For  
- example, [0,1,2,4,5,6,7] might        be        rotated        at        pivot        index 3   and become [4,5,6,7,0,1,2].Given the array nums after the possible rotation and an integer target, return the index of target if it is in nums, or -1   if it is not in nums.
- You must write an algorithm with O(log n) runtime complexity.
- Example 1:Input: `nums = [4,  5, 6, 7, 0, 1, 2], target = 0 Output:   4` 
- Example 2:Input: `nums = [4,  5, 6, 7, 0, 1, 2], target = 3` `Output:   -1` 
- Example 3:Input: `nums = [1], target = 0Output:   -1`

# Constraints:
- 1 <= nums . lengtth <= 5000 
- -104 <= nums [ i ] <= 104
- All values of nums are unique.
- nums is an ascending array that is possibly rotated.
- -104 <= target <= 104

# Approach
- First inilitaze left and right pointer with 0 and length of array
- Run while loop from left to right if left <= right
- Then calculate mid left+(right-left)//2
- check if mid value is equal target then return it's index `return mid`
- Other wise check if right element is less then mid and target also less then mid value then move the left pointer to `left=mid+1` else `right=mid-1`

# Time Complexity 
$$ O(log n) $$
# Space complexity
$$ O(1) $$

```python
def Bineary_search(nums,target):
    left,right=0,len(nums)-1
    while left<=right:
        mid=left+(right-left)//2
        if(nums[mid]==target):
            return mid
        elif(nums[left]<=nums[mid]):
            if(nums[left]<=target<=nums[mid]):
                right=mid-1
            else:
                left=mid+1
        else:
            if(nums[mid]<=target<=nums[right]):
                left=mid+1
            else:
                right=mid-1
    return -1
nums=[4,5,6,7,0,1,2]
target=5
print(Bineary_search(nums,target))
```

#### Question 3.
You  are  given  an  array prices where prices[i] is  the  price  of  a  given  stock  on  the ith day.Return the maximum profit you can achieve from this transaction. If you cannot achieve any profit, return 0.
- Example 1:Input: `prices = [7, 1, 5, 3, 6,  4]   Output:   5` Explanation:Buy on day 2 (price = 1) and sell on day 5 (price = 6), profit = 6-1 = 5. Note that buying on day 2 and selling on day 1 is not allowed because you must buy before you sell.
- Example 2:Input:`Prices = [7, 6, 4,  3, 1]   Output:   0` Explanation:In this case, no transactions are done and the max profit = 0.

# Constraints:
- 1 <= prices . length <= 105
- 0   <= prices [ i ] <= 104 


# Approach
- first inilitaze `max_profit=0`
- then inilitiaze `best_buy_price=prices[0]`
- then run the loop till ten of all array
- then check if current price grater then best day price then first calculate current profit
- After then modify max_profit with currnt and last max_profit
- then modify best_buy_prices with current min price
- at the end return max_profit
# Time complexity
$$ O(n) $$
# zSpace complexity
$$ O(1) $$

```python

def BuyAndSellStocks(prices):
    max_profit=0
    best_buy_price=prices[0]
    for curr in range(len(prices)):
        if prices[curr]>best_buy_price:
            curr_profit=prices[curr]-best_buy_price
            max_profit=max(curr_profit,max_profit)
        best_buy_price=min(best_buy_price,prices[curr])
    return max_profit
prices=[7,1,5,3,6,4]
print(BuyAndSellStocks(prices))
```

#### Question 4 
Given n non-negative integers representing an elevation map where the width of each bar is 1, compute how much water it can trap after raining.
- Example 1: Input: `height = [0, 1, 0,  2, 1, 0, 1, 3, 2, 1, 2, 1]` Output:   `6`  Explanation:The above elevation map (black section) is represented by a`rray [0,1,0,2,1,0,1,3,2,1,2,1]`. In this case, 6 units of rain water (blue section) are being trapped.
- Example 2:Input:` height = [4, 2, 0, 3, 2, 5]` Output:   `9`
# Constraints:
- n == height . length
- 1 <= n <= 2 * 104
- 0 <= height [ i ] < = 105

# Approach
- First of all inilitaze result variable with 0 value
- then inilitize left and right pointer `left=0 , right=len(height)-1`
- then get left max and right max
- then run the while loop `left<right`
- then check left_max < right_max if condition true then move left pointer next one step and find max of left poll then add value in resutl
- else move the right pointer and calculate max right and add water area in result 
- at the end return result
# time complexity
 $$ O(n)$$
 # space complexity
 $$ O(1) $$


```python
def trapRainWater(height):
    result=0
    left,right=0,len(height)-1
    left_max,right_max=height[left],height[right]
    while left<right:
        if left_max<right_max:
            left+=1
            left_max=max(left_max,height[left])
            result+=(left_max-height[left])
        else:
            right-=1
            right_max=max(right_max,height[right])
            result+=(right_max-height[right])
    return result
height=[4,2,0,3,2,5]
print(trapRainWater(height))
```
