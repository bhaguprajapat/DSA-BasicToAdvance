Valid Triangle
Given the three sides of a triangle a, b, and c. Check whether the triangle is valid or not.

Examples:

Input: a = 8, b = 15, c = 17
Output: Valid
Input: a = 3, b = 6, c = 9
Output: Invalid
Constraints:
1 <= a,b,c <= 106

```python
#User function Template for python3
# Approach: A triangle is valid if sum of its two sides is greater than the third side. If three sides are a, b and c, then three conditions should be met. 
# Time Complexity: O(1)
# Auxiliary Space: O(1)
class Solution:
    def checkValidity(self, a, b, c):
        if (a+b>c) and (a+c>b) and (b+c>a):
            return True
        else:
            return False
sl=Solution()
a = 8
b = 3
c = 3
print(sl.checkValidity(a,b,c))
```
