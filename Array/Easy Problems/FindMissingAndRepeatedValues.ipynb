#### Find Missing and Repeated Values
Leetcode problem <b> [https://leetcode.com/problems/find-missing-and-repeated-values/description/](https://leetcode.com/problems/find-missing-and-repeated-values/description/)</b>


```python
def findMissingAndRepeatedValues(grid):
    list1=[]
    result=[]
    for col in grid:
        for n in col:
            list1.append(n)
    list1=sorted(list1)
    if len(list1) !=list1[-1]:
        result.append(len(list1))
    for i in range(len(list1)-1):
        # check repeated value
        if list1[i]==list1[i+1]:
            result.insert(0,list1[i])
            # check missing valu
            if i+1  not in list1:
                result.append(i+1)
        return result
grid=[[1,3],[2,2]]
print(findMissingAndRepeatedValues(grid))
```
